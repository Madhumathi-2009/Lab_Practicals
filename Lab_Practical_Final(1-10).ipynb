# Use Conditinal Statements and Loops

```python
num = int(input("Enter a number: "))            # 1. Check whether a number is positive
if num >= 0:
    print("The number is Positive")
n = int(input("Enter a number: "))              # 2. Check whether a number is even or odd
print("The number is Even" if n % 2 == 0 else "The number is Odd")
marks = int(input("Enter student marks: "))     # 3. Assign grade based on marks
if marks > 90:
    print("Grade A")
elif marks > 60:
    print("Grade B")
elif marks > 35:
    print("Grade C")
else:
    print("Fail")
user = input("Username: ")                      # 4. Verify login credentials
pwd = input("Password: ")
if user == "Alice" and pwd == "1234":
    print("Login Successful")
else:
    print("Check your Credentials")
age = int(input("Enter your age: "))            # 5. Check voting eligibility
print("Eligible to Vote" if age >= 18 else "Not Eligible")
print("For Loop :")
for i in range(1, 6):                           # 6. Display numbers 1 to 5 using for loop
    print(i)
print("While Loop :")
i = 1                                           # 7. Display numbers 1 to 5 using while loop
while i <= 5:
    print(i)
    i += 1
print("Break Statement :")
for i in range(10):                             # 8. Demonstrate break statement
    if i == 6:
        break
    print(i)
print("Continue Statement :")
for i in range(10):                             # 9. Demonstrate continue statement
    if i == 5:
        continue
    print(i)
print("Pass Statement :")
for i in range(10):                             # 10. Demonstrate pass statement
    if i == 5:
        pass
    else:
        print(i)
day = int(input("Enter a day (1-7): "))         # 11. Display day using match-case
match day:
    case 1: print("Monday")
    case 2: print("Tuesday")
    case 3: print("Wednesday")
    case 4: print("Thursday")
    case 5: print("Friday")
    case 6: print("Saturday")
    case 7: print("Sunday")
    case _: print("Invalid Day")
```

# To Define and call user-defined functions

```python
# 1. Function without arguments and without return value
def greet():
    print("Welcome to Python")
greet()
# 2. Function with arguments (Addition)
def add(a, b):
    print("Sum =", a + b)

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
add(a, b)
# 3. Function without arguments that returns a value
def message():
    return "Welcome"

print(message())
# 4. Function with arguments that returns product
def product(a, b):
    return a * b
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print("Product =", product(a, b))
# 5. Function with default argument
def display_age(age=5):
    print("Age:", age)

display_age()
# 6. Function with keyword arguments
def student(name, age):
    print("Name:", name)
    print("Age :", age)
student(name="Alex", age=20)
# 7. Function with variable-length arguments (*args)
def total(*args):
    return sum(args)
print("Total =", total(56, 80, 20))
# 8. Lambda function to calculate square
square = lambda n: n * n
print("Square =", square(11))
# 9. Recursive function to calculate factorial
def fact(n):
    if n == 1:
        return 1
    return n * fact(n - 1)
n = int(input("Enter a number: "))
print("Factorial =", fact(n))
# 10. Nested function
def outer_function():
    def inner_function():
        print("Welcome to Python")
    inner_function()
outer_function()
```

# To work with Lists and List comprehensions

```python
print("-------------------------------------------------")
print("                  LIST OPERATIONS                ")
print("-------------------------------------------------")
print()
# Create a List containing Integer elements
l = [10, 15, 20, 25, 30, 35, 40, 45, 50, 60, 70]
print("Original list       :", l)
print()
print("First element       :", l[0])          # Accessing and displaying first element
print("Last element        :", l[-1])         # Accessing and displaying last element
print()
l.append(80)                                  # Adding element at the end of the list
print("After adding 80     :", l)
print()
l.insert(10, 65)                              # Inserting element at a specified position
print("After inserted 65   :", l)
print()
l.remove(30)                                  # Remove an element from the list
print("After removing 30   :", l)
print()
Length = len(l)                               # Length of the list
print("Length of the list  :", Length)
print()
print("Traverse of list :")                   # Traverse the list using for loop
for i in l:
    print(i, end=" ")
print()
print()
Square = [n*n for n in l]                     # Create a new list containing squares
print("Squares of the list elements :")
print(Square)
print()
Even = [n for n in l if n % 2 == 0]           # Create a list containing even numbers
print("Even       :", Even)
print()
Odd = [n for n in l if n % 2 != 0]            # Create a list containing odd numbers
print("Odd        :", Odd)
print()
Double = [n+n for n in l]                     # Create a new list by doubling each element
print("Double     :", Double)
print()
Matrix = [                                    # Create a nested list (matrix)
    [12, 15, 23],
    [27, 32, 35]
]                                             # Flatten the matrix using nested list comprehension
flatten = [element for row in Matrix for element in row]
print("Flattened  :", flatten)
```

# To demonstrate operations on Tuple and Sets

```python
print("-------------------------------------------------")
print("                  TUPLE OPERATIONS               ")
print("-------------------------------------------------", "\n")
t = (1, 2, 3, 1, 4, 5, 6, 1, 7, 8, 9, 10)         # Creating a tuple with integer elements
print("Original tuple      :", t ,"\n")           # Displaying the original tuple
print("First element       :", t[0])              # Accessing the first element
print("Last element        :", t[-1], "\n")       # Accessing the last element
l = len(t)                                        # Displaying the length of the tuple
print("Length of the tuple :", l, "\n")
print("Selected elements :")                      # Slicing and displaying selected elements
print(t[3:9])
print(t[:5])
print(t[5:])
print(t[::-1])
print()
print("Count of 1          :", t.count(1), "\n")  # Count the occurrence of a specific element
print("Index of 7          :", t.index(7), "\n")  # Find the index of a specified element
print("Traverse of tuple :")                      # Traverse the tuple using for loop
for i in t:
    print(i, end=" ")
print()
t1 = (20, 30, 40)                                 # Create another tuple
print("\nConcatenation of t and t1 :")            # Concatenate two tuples
print(t + t1, "\n")
print("-------------------------------------------------")
print("                  SET OPERATIONS                 ")
print("-------------------------------------------------", "\n")
s = {10, 20, 25, 30, 40, 45, 50}                   # Create a set with integer elements
print("Original Set          :", s)                # Display the original set
print()
s.add(55)                                          # Add a new element to the set
print("Adding the element 55 :", s)
print()
s.remove(25)                                       # Remove an element from the set
print("After Removed 25      :", s)
print()
length = len(s)                                    # Display the length of the set
print("Length of the Set     :", length)
print()
print("Traverse of the Set :")                     # Traverse the set using for loop
for i in s:
    print(i, end=" ")
print()
s1 = {5, 10, 15, 18, 30, 21, 48, 50, 60}           # Create another set
print()
print("METHOD 1")                                  # Set operations using methods
print("Union                :", s.union(s1))
print("Intersection         :", s.intersection(s1))
print("Difference           :", s.difference(s1))
print("Symmetric Difference :", s.symmetric_difference(s1), "\n")
print("METHOD 2")                                   # Set operations using operators
print("Union                :", s | s1)
print("Intersection         :", s & s1)
print("Difference           :", s - s1)
print("Symmetric Difference :", s ^ s1)
```

# To perform basic operations on Dictionaries

```python
print("-------------------------------------------------")
print("                    DICTIONARY                   ")
print("-------------------------------------------------")
# Creating a dictionary containing student details
student = {"Name" : "Rose", "Age" : 18, "Total Marks" : 519, "Department" : "CSE"}
print()
# Displaying the Original Dictionary
print("Original Dictionary         :",student, "\n")
# Accessing and displaying values of specific keys
print("Name                        :", student["Name"])
print("Department                  :",student["Department"], "\n")
# Adding a new key value pair
student["Percentage"] = 86.5
print("After adding percentage     :",student, "\n")
# Updating value of existing key
student["Department"] = "Cyber Security"
print("After updating value        :",student,"\n")
# Removing key value pairs
student.pop("Total Marks")
print("After Removing Marks        :",student,"\n")
# Length of the Dictionary
print("Length of the Dictinary     :",len(student),"\n")
# Display all keys
print("Keys                        :",student.keys(),"\n")
# Diplay all values
print("Values                      :",student.values(),"\n")
# Display all key value pairs
print("Items                       :",student.items(),"\n")
# Traverse dictionary using for loop
print("Traversing Dictionary :")
for i, j in student.items():
    print(i,":",j)
print()
# Check if specific key exists
if "roll" in student:
    print("Key 'roll' exists in dictionary")
else:
    print("key 'roll' does not exist","\n")
# Creating a copy of dictionary
print("Copy of dictionary          :",student.copy(),"\n")
# Creating another dictionary and clearing it
dict2 = {"a" : 1, "b" : 2, "c" : 3}
print("dict2 before clearing       :",dict2)
dict2.clear()
print("dict2 after clearing        :",dict2,"\n")
# Displaying all the updated dictionries
print("----------------------------------")
print("      Updated Dictionaries        ")
print("----------------------------------")
print("Original Dictionary :",student)
print("Copy of Dictionary  :",student.copy())
print("Cleared Dictionary  :",dict2)
```

# To use NumPy arrays and perform array operations

```python
import numpy as np                                  # Importing NumPy Library
a = np.array([10, 20, 30, 40, 50])                  # Creating two NumPy arrays with integer elements
b = np.array([2, 4, 6, 8, 10])
print("Array 1 :", a)                               # Displaying both arrays
print("Array 2 :", a,"\n")
print("Addition      :", a + b)                     # Element-wise operations
print("Subtraction   :", a - b)
print("Multiplication:", a * b)
print("Division      :", a / b,"\n")
print("First element of Array 1:", a[0])            # Accessing first element using indexing
print("Slicing arr1[1:4]       :", a[1:4], "\n")    # Extracting portion using slicing
c = np.array([1, 2, 3, 4, 5, 6])                    # Creating another array and reshape to 2D
c_reshaped = c.reshape(2, 3)
print("Reshaped 2D Array:\n", c_reshaped)           # Displaying reshaped array with shape, dimension, size
print("Shape    :", c_reshaped.shape)               # Shape of the array
print("Dimension:", c_reshaped.ndim)                # Dimension of the array
print("Size     :", c_reshaped.size,"\n")           # Size of the array
print("Sum of Array 1 :", np.sum(a))                # Aggregation operation
print("Mean of Array 1:", np.mean(a))
print("Max of Arrat 1 :", np.max(a))
print("Min of Array 1 :", np.min(a),"\n")
print("Square root of Array 2:", np.sqrt(b))        # Universal functions - sqrt and exp       
print("Exponential of Array 2:", np.exp(b),"\n")
arr_unsorted = np.array([50, 10, 40, 20, 30])       # Sort the elements
sorted_arr = np.sort(arr_unsorted)
print("Original array    :", arr_unsorted)
print("Sorted array      :", sorted_arr)
concatenated = np.concatenate((a, b))               # Concatenate two arrays       
print("Concatenated Array:", concatenated)
```

# To perform indexing, slicing and reshaping in NumPy

```python
import numpy as np                                # Importing NumPy library
a = np.array([5, 10, 15, 20, 25, 30])             # Creating a 1D NumPy array 
print("1D Array:", a,"\n")                        # Displaying the  1D array
print("Indexing:")                                # Accessing specific elements using indexing
print("Element at index 0:", a[0])
print("Element at index 3:", a[3])
print("Last element      :", a[-1],"\n")
print("Slicing:")                                 # Extract subsets using slicing
print("Elements from index 1 to 4:", a[1:5])
print("First 3 elements          :", a[:3])
print("Last 2 elements           :", a[-2:])
print("Every 2nd element         :", a[::2],"\n")
b = np.array([[1, 2, 3],                          # Creating a 2D NumPy array
              [4, 5, 6], 
              [7, 8, 9]])
print("2D Array:\n",b,"\n")                       # Displaying the 2D array
print("Accessing 2D elements:")                   # Accessing individual elements from 2D array
print("Element at row 0, col 1:", b[0, 1])
print("Element at row 2, col 2:", b[2, 2],"\n")
print("Slicing 2D array:")                        # Extracting rows and columns from 2D array
print("First row    :",b[0, :])
print("Second column:",b[:, 1])
print("Sub-matrix [0:2, 1:3]:\n",b[0:2, 1:3],"\n")
arr_reshape = np.array([1, 2, 3, 4, 5, 6])        # Creating another 1D array for reshaping
print("Original 1D array for reshaping:", arr_reshape,"\n")
matrix_2x3 = arr_reshape.reshape(2, 3)            # Reshaping into 2x3 matrix
print("Reshaped to 2x3 matrix:\n", matrix_2x3,"\n")
matrix_3x2 = arr_reshape.reshape(3, 2)            # Reshaping into 3x2 matrix
print("Reshaped to 3x2 matrix:\n", matrix_3x2,"\n")
flattened = matrix_2x3.flatten()                  # Converting reshaped array back to 1D using flatten()
print("Flattened back to 1D:", flattened,"\n")
print("Array Attributes for 2x3 matrix:")         # Displaying array attributes
print("Shape     :", matrix_2x3.shape)            
print("Dimensions:", matrix_2x3.ndim)
print("Size      :", matrix_2x3.size)
print("Data type :", matrix_2x3.dtype)
```

# To use statistical functions with Numpy

```python
print("----------------------------------")
print("       STATISTICAL FUNCTIONS      ")
print("----------------------------------")
import numpy as np                                          # Importing NumPy library
data = np.array([12, 15, 18, 22, 25, 28, 30, 35, 40, 45])   # Creating a NumPy array 
print("Array elements    :", data)                          # Displaying the array elements
print("Sum of elements   :", np.sum(data))                  # Sum of all elements
print("Mean              :", np.mean(data))                 # Mean of the array
print("Median            :", np.median(data))               # Median of the array
print("Standard Deviation:", np.std(data))                  # Standard deviation of the array
print("Variance          :", np.var(data))                  # Variance of the array
print("Minimum value     :", np.min(data))                  # Minimum value            
print("Maximum value     :", np.max(data))                  # Maximum value
print("Index of Minimum value:", np.argmin(data))           # Indices of min value
print("Index of Maximum value:", np.argmax(data),"\n")      # Indices of max value
data_range = np.max(data) - np.min(data)                    # Range of the array
print("Range:", data_range,"\n")
# 25th, 50th, and 75th percentiles
print("25th Percentile         :", np.percentile(data, 25))
print("50th Percentile (Median):", np.percentile(data, 50))
print("75th Percentile         :", np.percentile(data, 75),"\n")
# Cumulative sum
print("Cumulative Sum :", np.cumsum(data))
# Sort the array
sorted_data = np.sort(data)
print("Sorted Array :", sorted_data)  
```
