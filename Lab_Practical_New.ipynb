# Code 2

```python
# 1. Check whether a number is positive
num = int(input("Enter a number: "))
if num >= 0:
    print("The number is Positive")

# 2. Check whether a number is even or odd
n = int(input("Enter a number: "))
print("The number is Even" if n % 2 == 0 else "The number is Odd")

# 3. Assign grade based on marks
marks = int(input("Enter student marks: "))
if marks > 90:
    print("Grade A")
elif marks > 60:
    print("Grade B")
elif marks > 35:
    print("Grade C")
else:
    print("Fail")

# 4. Verify login credentials
user = input("Username: ")
pwd = input("Password: ")
if user == "Alice" and pwd == "1234":
    print("Login Successful")
else:
    print("Check your Credentials")

# 5. Check voting eligibility
age = int(input("Enter your age: "))
print("Eligible to Vote" if age >= 18 else "Not Eligible")

# 6. Display numbers 1 to 5 using for loop
print("\nFor Loop:")
for i in range(1, 6):
    print(i)

# 7. Display numbers 1 to 5 using while loop
print("\nWhile Loop:")
i = 1
while i <= 5:
    print(i)
    i += 1

# 8. Demonstrate break statement
print("\nBreak Statement:")
for i in range(10):
    if i == 6:
        break
    print(i)

# 9. Demonstrate continue statement
print("\nContinue Statement:")
for i in range(10):
    if i == 5:
        continue
    print(i)

# 10. Demonstrate pass statement
print("\nPass Statement:")
for i in range(10):
    if i == 5:
        pass
    else:
        print(i)

# 11. Display day using match-case
day = int(input("\nEnter a day (1-7): "))
match day:
    case 1: print("Monday")
    case 2: print("Tuesday")
    case 3: print("Wednesday")
    case 4: print("Thursday")
    case 5: print("Friday")
    case 6: print("Saturday")
    case 7: print("Sunday")
    case _: print("Invalid Day")
```

# Code 3

```python
# 1. Function without arguments and without return value
def greet():
    print("Welcome to Python")
greet()

# 2. Function with arguments (Addition)
def add(a, b):
    print("Sum =", a + b)

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
add(a, b)

# 3. Function without arguments that returns a value
def message():
    return "Welcome"

print(message())

# 4. Function with arguments that returns product
def product(a, b):
    return a * b

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print("Product =", product(a, b))

# 5. Function with default argument
def display_age(age=5):
    print("Age:", age)

display_age()

# 6. Function with keyword arguments
def student(name, age):
    print("Name:", name)
    print("Age :", age)

student(name="Alex", age=20)

# 7. Function with variable-length arguments (*args)
def total(*args):
    return sum(args)

print("Total =", total(56, 80, 20))

# 8. Lambda function to calculate square
square = lambda n: n * n
print("Square =", square(11))

# 9. Recursive function to calculate factorial
def fact(n):
    if n == 1:
        return 1
    return n * fact(n - 1)

n = int(input("Enter a number: "))
print("Factorial =", fact(n))

# 10. Nested function
def outer_function():
    def inner_function():
        print("Welcome to Python")
    inner_function()

outer_function()
```

# Code 4

```python
print("-------------------------------------------------")
print("              LIST OPERATIONS             ")
print("-------------------------------------------------")
print()


# Create a List containing Integer elements
l = [10, 15, 20, 25, 30, 35, 40, 45, 50, 60, 70]
print("Original list       :",l)
print()

# Accessong and displaying girst and last element
print("First element       :", l[0])
print("Last element        :", l[-1])
print()

# Adding element at the end of the list
l.append(80)
print("After adding 80     :",l)
print()

# Insering element at a specified position
l.insert(10,65)
print("After inserted 65   :",l)
print()

# Remove an element from the list
l.remove(30)
print("After removing 30   :",l)
print()

# Length of the list
Length = len(l)
print("Length of the list  :",Length)
print()

# Traverse the list using for
print("Traverse of list :")
for i in l:
    print(i,end=" ")
print()
print()

# Create a new list containing the squares of all elements using list comprehension
print("Squares of the list elements :")
Square = [n*n for n in l]
print(Square)
print()

# Create a list containing even numbers using list comprehensions
Even = [n for n in l if n % 2 == 0]
print("Even       :",Even)
print()

# Create a list containing odd numbers using list comprehensions
Odd = [n for n in l if n % 2 != 0]
print("Odd        :",Odd)
print()

# Create a new list by doubling each element using list comprehension
Double = [n+n for n in l]
print("Double     :",Double)
print()

# Create a nested list (matrix) and flatten it into a single list using nested list comprehension
Matrix = [
    [12, 15, 23],
    [27, 32, 35]
]
flatten = [element for row in Matrix for element in row]
print("Flattened  :",flatten)
```

```python
print("-------------------------------------------------")
print("                DICTIONARY             ")
print("-------------------------------------------------")
print()

# Creating a dictionary containing student details
student = {"Name" : "Rose", "Age" : 18, "Total Marks" : 519, "Department" : "CSE"}
print()

# Displaying the Original Dictionary
print("Original Dictionary         :",student)
print()

# Accessing and displaying values of specific keys
print("Name                        :", student["Name"])
print("Department                  :",student["Department"])
print()

# Adding a new key value pair
student["Percentage"] = 86.5
print("After adding percentage     :",student)
print()

# Updating value of existing key
student["Department"] = "Cyber Security"
print("After updating value        :",student)
print()

# Removing key value pairs
student.pop("Total Marks")
print("After Removing Marks        :",student)
print()

# Length of the Dictionary
print("Length of the Dictinary     :",len(student))
print()

# Display all keys
print("Keys                        :",student.keys())
print()

# Diplay all values
print("Values                      :",student.values())
print()

# Display all key value pairs
print("Items                       :",student.items())
print()

# Traverse dictionary using for loop
print("Traversing Dictionary :")
for i, j in student.items():
    print(i,":",j)
print()

# Check if specific key exists
if "roll" in student:
    print("Key 'roll' exists in dictionary")
else:
    print("key 'roll' does not exist")
print()

# Creating a copy of dictionary
print("Copy of dictionary          :",student.copy())
print()

# Creating another dictionary and clearing it
dict2 = {"a" : 1, "b" : 2, "c" : 3}
print("dict2 before clearing       :",dict2)
dict2.clear()
print("dict2 after clearing        :",dict2)
print()

# Displaying all the updated dictionries
print("----------------------------------")
print("      Updated Dictionaries        ")
print("----------------------------------")
print("Original Dictionary :",student)
print("Copy of Dictionary  :",student.copy())
print("Cleared Dictionary  :",dict2)
```
