# Use Conditinal Statements and Loops

1. To check whether a number positive using if statement

```python
Number = int(input("Enter a number :"))
if Number >= 0:
    print("The Number is Positive")
```

2. To determine whether it is even or odd using if-else statement

```python
N = int(input("Enter a Number :"))
if N%2==0:
    print("The Number is Even")
else:
    print("The Number is Odd")
```

3. To assign a grade according to the Student marks

```python
M = int(input("Enter Student Marks :"))
if M > 90:
    print("Grade A")
elif M > 60:
    print("Grade B")
elif M > 35:
    print("Grade C")
else:
    print("Fail")
```

4. To Verify the login credentials using nested-if-else statement

```python
a = input("Username :")
b = input("Password :")
if a == "Alice":
    if b == "1234":
        print("Login Successfully")
    else:
        print("Check your Credentials")
else:
    print("Check your Credentials")
```

5. To check whether the user is eligible for voting or not Using one-line-if-else statement

```python
A = int(input())
s = "Eligible to vote" if A >= 18 else "Not Eligible"
print(s)
```

6. To Display numbers from 1 to 5 using for loop

```python
for i in range(1,6,1):
    print(i)
```

7. To Display numbers from 1 to 5 using while loop

```python
i = 1 
while(i<6):
    print(i)
    i = i + 1
```

8. To Demonstrate the break statement by terminating the loop when the value becomes 5

```python
for i in range(10):
    if i == 6:
        break
    print(i)
```

9. To Demonstrate the continue statement by skipping the value 5 during loop execution

```python
for i in range(10):
    if i == 5:
        continue
    print(i)
```

10. To demonstrate the pass statement as a placeholder while continuing loop execution

```python
for i in range(10):
    if i == 5:
        pass
    else:
        print(i)
```

# To Define and call user-defined functions

1. Define a function without arguements and without a return value, then call it

```python
def greet():
    print("Welcome to Python")
greet()
```

2. Define a function with arguements to perform addition and call it with two values

```python
def add(a,b):
    print(a+b)
a = int(input())
b = int(input())
add(a,b)
```

3. Function without arguements that returns a value

```python
def greet():
    return "Welcome"
a = greet()
print(a)
```

4. Function with arguements that returns product

```python
def product(a,b):
    return a*b
a = int(input())
b = int(input())
product(a,b)
```

5. Function with default arguement

```python
def num(a=5):
    print("age:",a)
num()
```

6. Function with keyword arguements

```python
def student(Name, Age):
    print(Name, Age)
student(Name = "Alex", Age = 20)
```

7. Function with variable length arguements(*args)

```python
def total(*args):
    total=0
    for i in args:
        total+=i
    return total
total(56,80,20)
```

8. Lambda Function to calculate the sqaure of a number

```python
Sq = lambda n: n*n
print(Sq(11))
```

9. Recursive function to calculate the factorial of a number

```python
def fact(n):
    if n==1:
        return 1
    return n*fact(n-1)
n = int(input())
print(fact(n))
```

10. Nested function and call inner function from the outer function

```python
def Outer_Function():
    def Inner_Function():
        print("Welcome to Python")
    Inner_Function()
Outer_Function()
```

# To work with Lists and List comprehensions

```python
print("-------------------------------------------------")
print("              LIST OPERATIONS             ")
print("-------------------------------------------------")
print()


# Create a List containing Integer elements
l = [10, 15, 20, 25, 30, 35, 40, 45, 50, 60, 70]
print("Original list :",l)
print()

# Accessong and displaying girst and last element
print("First element :", l[0])         
print("Last element :", l[-1])          
print()

# Adding element at the end of the list
l.append(80)
print("After adding 80 :",l)
print()

# Insering element at a specified position
l.insert(10,65)
print("After inserted 65 :",l)
print()

# Remove an element from the list
l.remove(30)
print("After removing 30 :",l)
print()

# Length of the list
Length = len(l)
print("Length of the list :",Length)
print()

# Traverse the list using for 
print("Traverse of list :")
for i in l:
    print(i,end=" ")
print()
print()

# Create a new list containing the squares of all elements using list comprehension
print("Squares of the list elements :")
Square = [n*n for n in l]
print(Square)
print()

# Create a list containing even numbers using list comprehensions
Even = [n for n in l if n % 2 == 0]
print("Even :",Even)
print()

# Create a list containing odd numbers using list comprehensions
Odd = [n for n in l if n % 2 != 0]
print("Odd :",Odd)
print()

# Create a new list by doubling each element using list comprehension
Double = [n+n for n in l]
print("Double :",Double)
print()

# Create a nested list (matrix) and flatten it into a single list using nested list comprehension
Matrix = [
    [12, 15, 23],
    [27, 32, 35]
]
flatten = [element for row in Matrix for element in row]
print("Flattened :",flatten)
```

# To demonstrate operations on Tuple and Sets

```python
print("-------------------------------------------------")
print("              TUPLE OPERATIONS            ")
print("-------------------------------------------------")
print()


# Creating a tuple with integer elements
t = (1, 2, 3, 1, 4, 5, 6, 1, 7, 8, 9, 10)

# Displaying the original tuple
print("Original tuple :",t)
print()

# Accessing and displaying the first and last element of a tuple
print("First element :",t[0])
print("Last element :",t[-1])
print()

# Displaying the length of the tuple
l = len(t)
print("Length of the tuple :",l)
print()

# Slicing and displayong the selected elements
print("Selected elements :")
print(t[3:9])
print(t[:5])
print(t[5:])
print(t[::-1])
print()

# To count the occurance of a specific element
print("Count of 1 :",t.count(1))
print()

# To find the index of a specified element
print("Index of 7 :",t.index(7))
print()

# Traverse of tuple using for loop
print("Tranverse of tuple :")
for i in t:
    print(i,end=" ")
print()

# Concanate two tuples
t1 = (20, 30, 40)
print("\nConcatenation of t and t1 :")
print(t+t1)
print()


print("-------------------------------------------------")
print("              SET OPERATIONS            ")
print("-------------------------------------------------")
print()


# Create a set with integer elements
s = {10, 20, 25, 30, 40, 45, 50}

# Displaying the Original set
print("Original Set :",s)
print()

# Adding new element to the set
s.add(55)
print("Adding the element 55 :",s)
print()

# Removing an element from the set
s.remove(25)
print(s)
print()

# Length of set 
length = len(s)
print("Length of the Set :",length)
print()

# Traverse of the Set
print("Traverse of the Set :")
for i in s:
    print(i,end=" ")
print()
print()

# Creating another Set with integer elements
s1 = {5, 10, 15, 18, 30, 21, 48, 50, 60}

# To display the union, intersection, difference and symmetric difference of two sets
print("METHOD 1")
print("Union :",s.union(s1))
print("Intersection :",s.intersection(s1))
print("Difference :",s.difference(s1))
print("Symmetric Difference :",s.symmetric_difference(s1))
print()
print("METHOD 2")
print("Union :",s|s1)
print("Intersection :",s&s1)
print("Difference :",s-s1)
print("Symmetric Difference :",s^s1)
```
